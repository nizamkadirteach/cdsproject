```python
import sys
sys.path.append('../src')
import data_loader
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="darkgrid")
```

## 1. Data Collection & Visualization

```python
df = data_loader.fetch_crypto_data(ticker="BTC-USD", start_date="2023-01-01")
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

plt.figure(figsize=(14, 7))
plt.plot(df['Close'], label='BTC Close Price')
plt.title('Bitcoin Price History (2023-Present)', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()
```

## 2. Naive Model Implementation
We use a simple **Persistence Model** (Naive Forecast) where the prediction for tomorrow is simply today's price ($ \hat{y}_{t+1} = y_t $).

```python
df['Naive_Pred'] = df['Close'].shift(1)
df.dropna(inplace=True)

# Calculate Error Metrics
mse = ((df['Close'] - df['Naive_Pred']) ** 2).mean()
mae = (df['Close'] - df['Naive_Pred']).abs().mean()

print(f"Naive Model MSE: {mse:.2f}")
print(f"Naive Model MAE: {mae:.2f}")
```

```python
plt.figure(figsize=(14, 7))
plt.plot(df.index[-100:], df['Close'][-100:], label='Actual', linewidth=2)
plt.plot(df.index[-100:], df['Naive_Pred'][-100:], label='Naive Forecast', linestyle='--', color='red')
plt.title('Naive Model vs Actual Price (Last 100 Days)', fontsize=16)
plt.legend()
plt.show()
```
